```python
print("Mohd. Affan Ansari", "UIN: 231A023\n")
text_lower=text.lower()
print(text_lower)
```

```python
print("Mohd. Affan Ansari", "UIN: 231A023\n")
import re
pattern=r'https?://www\.\S+|www\.\S+|\S+\@\S+|\@\S+' #we remove url's, emails & twiter handle through this code.
text_url=re.sub(pattern," ",text_lower)
print(text_url)
```

```python
pip install emoji
```

```python
import emoji
text_emoji = emoji.replace_emoji(text_url, replace="")
print(text_emoji)
```

```python
pattern=r'\d+\S+'
text_numeric=re.sub(pattern," ",text_emoji)
print(text_numeric)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk import word_tokenize
tokenized=word_tokenize (text_emoji)
print("Tokenized words:", tokenized)
```

```python
print("Mohd. Affan Ansari", "UIN: 231A023\n")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("Mohd. Affan Ansari", "UIN: 231A023\n")
print("With punctuation:", tokenized)
print()
text_punct=[]
for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)
print("Without punctuation:", text_punct)
```

```python
#removing stopwords
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
```

```python
print("Mohd. Affan Ansari", "UIN: 231A023\n")
print("With stopwords (after punctuation removal):")
print(text_punct)
print()
text_stopword = []
for word in text_punct:
    if word.lower() not in stop_words:
        text_stopword.append(word)
print("Without stopwords:")
print(text_stopword)
```

```python
#stemming words
from nltk.stem import PorterStemmer
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in text_stopword]
print("Stemmed words:", stemmed_words)
```

```python
print("Words before stemming:\n", text_stopword)
print("\nWords after stemming:\n", stemmed_words)
```

```python
#lemitaization of words
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in text_stopword]
print("Words before lemmatization:", text_stopword)
print("\nWords after lemmatization:", lemmatized_words)
```

```python
print("\nWords after lemmatization:", lemmatized_words)
print("\nWords after stemming:", stemmed_words)
```

```python
print("Comparison of Stemming vs. Lemmatization:")
print("-----------------------------------------")
print("Original (without stopwords): ", text_stopword[:10]) # Showing first 10 for brevity
print("Stemmed (first 10 words):     ", stemmed_words[:10])
print("Lemmatized (first 10 words):  ", lemmatized_words[:10])

print("\nFull stemmed words list length:", len(stemmed_words))
print("Full lemmatized words list length:", len(lemmatized_words))
```
